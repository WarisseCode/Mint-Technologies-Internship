# Milestone 2: Feature Engineering

### Step 1: Library Imports and Data Loading

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# Load the dataset (adjust the path if necessary)
df = pd.read_csv('data/kc_house_data.csv')

# Display the first few rows to verify loading
df.head()
```

### Step 2: Categorical Encoding

```python
# Example with the 'zipcode' column or any other categorical column of your choice
if 'zipcode' in df.columns:
    df = pd.get_dummies(df, columns=['zipcode'], drop_first=True, dtype=int)

print("Dimensions after encoding:", df.shape)
```

One-Hot Encoding transforms a categorical variable (like zipcode) into multiple binary columns (0 or 1). This is crucial because categorical values often have arbitrary numeric representations; keeping them as a single numerical feature would mislead the machine learning model into assuming a mathematical relationship or false ordering between categories where none exists.

### Step 3: Feature Creation based on Business Logic

```python
# 1. Extract the sale year to calculate the age of the house
if 'date' in df.columns:
    df['sale_year'] = pd.to_datetime(df['date']).dt.year
    df['house_age'] = df['sale_year'] - df['yr_built']

# 2. Create the recommended 'price_per_sqft' variable
if 'price' in df.columns and 'sqft_living' in df.columns:
    df['price_per_sqft'] = df['price'] / df['sqft_living']

df[['house_age', 'price_per_sqft']].head()
```

These features were created to capture key domain-specific insights:
* **House Age (`house_age`)**: The age of a property is a critical factor influencing buyer decisions and property value, as newer homes generally command higher prices or lower maintenance costs.
* **Price per Square Foot (`price_per_sqft`)**: This metric standardizes property value, reflecting the intrinsic worth of the living space regardless of the overall size of the house, allowing for a more accurate comparison across different properties.

### Step 4: Feature Scaling

```python
# Initialize the scaler
scaler = StandardScaler()

# Select the continuous numerical columns to normalize
numerical_cols = ['sqft_living', 'sqft_lot']

for col in numerical_cols:
    if col in df.columns:
        # Apply Standard Scaler
        df[col + '_scaled'] = scaler.fit_transform(df[[col]])

# Verify the scaled columns
df[[col + '_scaled' for col in numerical_cols if col in df.columns]].head()
```

### Step 5: Saving the Final Dataset

```python
# Save the dataset ready for modeling
df.to_csv('data/engineered_data.csv', index=False)
print("Dataset successfully saved to data/engineered_data.csv!")
```
